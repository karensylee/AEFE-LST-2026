## 3. Verify & List Output Files

```python
import os

print("Files saved to Google Drive:")
print("=" * 50)

model_dir = os.path.join(OUTPUT_DIR, 'xgb', 'BLAM-ALL-CLOUDY')
if os.path.exists(model_dir):
    print("\nBLAM-ALL-CLOUDY Model Files:")
    for f in os.listdir(model_dir):
        fpath = os.path.join(model_dir, f)
        size_mb = os.path.getsize(fpath) / 1e6
        print(f"  {f}: {size_mb:.2f} MB")

print("\nSHAP Analysis:")
shap_dir = os.path.join(OUTPUT_DIR, 'shap_analysis_blam_all')
if os.path.exists(shap_dir):
    for f in os.listdir(shap_dir):
        if 'BLAM-ALL-CLOUDY' in f:
            print(f"  {f}")
else:
    print("  SHAP output directory not found")
```
